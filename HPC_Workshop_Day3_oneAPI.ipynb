```python
import os, sys, re, json, time, glob, shutil, platform, subprocess, textwrap
from pathlib import Path

# ---- Workspace on the VM's local disk (NO Google Drive needed) ----
BASE = "/content" if os.path.isdir("/content") else os.getcwd()
WORKSPACE = os.path.join(BASE, "hpc_day03")
for sub in ("bin", "logs", "results"):
    os.makedirs(os.path.join(WORKSPACE, sub), exist_ok=True)
os.chdir(WORKSPACE)
print(f"[+] Workspace ready at: {WORKSPACE}   (temporary - download your results at the end)")

# ---- Small helpers ----
def sh(cmd, show=True):
    """Run a shell command in the workspace; return its text output."""
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True, cwd=WORKSPACE)
    out = (r.stdout or "") + (r.stderr or "")
    if show and out.strip():
        print(out.rstrip())
    return out

TF_NOISE = re.compile(r"^(I0000|W0000|E0000)|absl::InitializeLog|WARNING: All log messages")
def show_log(path, tail=None):
    """Print a job log, hiding TensorFlow's start-up chatter (and report how many ranks had oneDNN on)."""
    txt = Path(path).read_text()
    n_onednn = txt.count("oneDNN custom operations are on")
    lines = [l for l in txt.splitlines() if not TF_NOISE.search(l)]
    if tail:
        lines = lines[-tail:]
    print("\n".join(lines))
    if n_onednn:
        print(f"\n[info] oneDNN optimizations were reported ON by {n_onednn} TensorFlow process(es).")

cpu_cores = os.cpu_count() or 1
print(f"Python {platform.python_version()} | {platform.system()} {platform.release()} | logical CPU cores: {cpu_cores}")
```

```python
%%writefile bench_onednn.py
"""bench_onednn.py - micro-benchmarks that show what oneDNN/MKL-style optimisation does inside TensorFlow.
Run it in a FRESH process with TF_ENABLE_ONEDNN_OPTS=0 or 1 (the flag is read when TensorFlow is imported)."""
import os, sys, json, time
import numpy as np
import tensorflow as tf

use_bf16 = os.environ.get("BENCH_BF16") == "1"
if use_bf16:
    tf.keras.mixed_precision.set_global_policy("mixed_bfloat16")
tf.keras.utils.set_random_seed(0)

def best_ms(fn, reps=5, warm=2):
    for _ in range(warm):
        fn()
    ts = []
    for _ in range(reps):
        t0 = time.perf_counter(); fn(); ts.append(time.perf_counter() - t0)
    return min(ts) * 1000

# 1) dense matrix multiply  (the core of every fully-connected layer)
N = int(os.environ.get("BENCH_MATMUL_N", "1536"))
a = tf.random.normal((N, N)); b = tf.random.normal((N, N))
dt = tf.bfloat16 if use_bf16 else tf.float32
a, b = tf.cast(a, dt), tf.cast(b, dt)

@tf.function
def mm(x, y):
    return tf.matmul(x, y)
matmul_ms = best_ms(lambda: mm(a, b).numpy())

# 2) small CNN training step (convolutions + backprop)
bs = int(os.environ.get("BENCH_BATCH", "64"))
model = tf.keras.Sequential([
    tf.keras.Input((64, 64, 3)),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(10, dtype="float32"),
])
opt = tf.keras.optimizers.Adam(1e-3)
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
x = tf.random.normal((bs, 64, 64, 3)); y = tf.random.uniform((bs,), maxval=10, dtype=tf.int32)

@tf.function
def train_step(x, y):
    with tf.GradientTape() as tape:
        loss = loss_fn(y, model(x, training=True))
    grads = tape.gradient(loss, model.trainable_variables)
    opt.apply_gradients(zip(grads, model.trainable_variables))
    return loss
cnn_ms = best_ms(lambda: train_step(x, y).numpy(), reps=5, warm=3)

print("RESULT:" + json.dumps({"matmul_ms": round(matmul_ms, 2), "cnn_step_ms": round(cnn_ms, 2),
                              "matmul_n": N, "batch": bs, "bf16": use_bf16,
                              "onednn_flag": os.environ.get("TF_ENABLE_ONEDNN_OPTS", "default"),
                              "tf": tf.__version__}))
```

```python
def run_bench(env_extra):
    """Run bench_onednn.py in a fresh process with the given environment."""
    env = {**os.environ, "TF_CPP_MIN_LOG_LEVEL": "0",
           "CUDA_VISIBLE_DEVICES": "-1",          # force CPU so we measure oneDNN, not a GPU
           **env_extra}
    r = subprocess.run([sys.executable, "bench_onednn.py"], env=env, capture_output=True, text=True,
                       cwd=WORKSPACE, timeout=1200)
    line = next((l for l in r.stdout.splitlines() if l.startswith("RESULT:")), None)
    if line is None:
        raise RuntimeError("benchmark failed:\n" + r.stderr[-1500:])
    res = json.loads(line[len("RESULT:"):])
    res["oneDNN message seen"] = "oneDNN custom operations are on" in r.stderr
    return res

configs = [("oneDNN OFF", {"TF_ENABLE_ONEDNN_OPTS": "0"}),
           ("oneDNN ON",  {"TF_ENABLE_ONEDNN_OPTS": "1"})]
if {"avx512_bf16", "amx_bf16"} & flags:            # hardware has native bfloat16 support
    configs.append(("oneDNN ON + bfloat16", {"TF_ENABLE_ONEDNN_OPTS": "1", "BENCH_BF16": "1"}))
else:
    print("[i] This CPU has no native bfloat16 units (AVX512-BF16 / AMX) -> skipping the bfloat16 test.\n")

onednn_results = {}
for name, env in configs:
    print(f"[*] Benchmarking: {name} ...")
    onednn_results[name] = run_bench(env)

base = onednn_results["oneDNN OFF"]
print("\n" + "=" * 78)
print(f"{'Configuration':<26}{'MatMul (ms)':>13}{'CNN train step (ms)':>22}{'Speedup (CNN)':>15}")
print("-" * 78)
for name, r in onednn_results.items():
    print(f"{name:<26}{r['matmul_ms']:>13.1f}{r['cnn_step_ms']:>22.1f}{base['cnn_step_ms'] / r['cnn_step_ms']:>14.2f}x")
print("=" * 78)
print(f"oneDNN banner seen in TensorFlow's log -> OFF run: {onednn_results['oneDNN OFF']['oneDNN message seen']}, "
      f"ON run: {onednn_results['oneDNN ON']['oneDNN message seen']}")

import matplotlib.pyplot as plt
names = list(onednn_results)
fig, axes = plt.subplots(1, 2, figsize=(11, 3.6))
for ax, key, title in zip(axes, ("matmul_ms", "cnn_step_ms"), ("Matrix multiply 1536x1536", "Small CNN training step")):
    vals = [onednn_results[n][key] for n in names]
    bars = ax.bar(names, vals, color=["#9e9e9e", "#0071c5", "#00a3e0"][:len(names)])
    ax.set_title(title); ax.set_ylabel("milliseconds (lower is better)")
    ax.tick_params(axis="x", rotation=12)
    for b, v in zip(bars, vals):
        ax.text(b.get_x() + b.get_width() / 2, v, f"{v:.0f}", ha="center", va="bottom")
plt.tight_layout(); plt.show()
```

### Extension Activity 1.1: Inspecting Cache Line Sizes and CPU Topology
We will read from the virtual files in `/sys/devices/system/cpu/cpu*/cache/` to inspect the cache architecture of the current VM, including the cache-line size, cache types, and levels.

```python
import os
from pathlib import Path

def inspect_cpu_cache():
    cache_dir = Path("/sys/devices/system/cpu/cpu0/cache")
    if not cache_dir.exists():
        print("[-] Cache information directory not found. This might not be a standard Linux environment.")
        return

    print("=" * 60)
    print(" CPU CACHE HIERARCHY & PROPERTIES (CPU 0)")
    print("=" * 60)

    # Iterate through the index folders (index0, index1, etc.) representing different cache levels
    for index_path in sorted(cache_dir.glob("index*")):
        try:
            level = (index_path / "level").read_text().strip()
            type_ = (index_path / "type").read_text().strip()
            size = (index_path / "size").read_text().strip()
            coherency_line_size = (index_path / "coherency_line_size").read_text().strip()
            shared_cpu_list = (index_path / "shared_cpu_list").read_text().strip()

            print(f"• L{level} {type_:<10} | Size: {size:<6} | Cache Line Size: {coherency_line_size} bytes | Shared CPUs: {shared_cpu_list}")
        except Exception as e:
            print(f"[-] Error reading {index_path.name}: {e}")
    print("=" * 60)

inspect_cpu_cache()
```

### Extension Activity 2.1: Micro-Benchmarking Convolutional Channels
Hardware architectures perform better when memory configurations are aligned to typical word and vector sizes. Let's write a benchmark script `bench_channels.py` to test how oneDNN speeds up power-of-two channel sizes (64, 128) versus non-power-of-two channel sizes (61, 123).

```python
%%writefile bench_channels.py
import os, sys, json, time
import tensorflow as tf

tf.keras.utils.set_random_seed(0)

def measure_conv_time(channels, reps=15, warm=3):
    # Synthetic batch of images (batch_size=64, height=64, width=64, input_channels=3)
    x = tf.random.normal((64, 64, 64, 3))
    # Create a 2D convolution layer with the requested number of output channels
    conv_layer = tf.keras.layers.Conv2D(channels, 3, activation="relu")

    @tf.function
    def run_conv(inputs):
        return conv_layer(inputs)

    # Warm-up
    for _ in range(warm):
        _ = run_conv(x).numpy()

    ts = []
    for _ in range(reps):
        t0 = time.perf_counter()
        res = run_conv(x).numpy()
        ts.append(time.perf_counter() - t0)

    return min(ts) * 1000  # Best execution time in milliseconds

# Benchmark both a power-of-two configuration and an unaligned one
power_of_two_channels = 128
non_power_of_two_channels = 123

pot_ms = measure_conv_time(power_of_two_channels)
npot_ms = measure_conv_time(non_power_of_two_channels)

print("RESULT:" + json.dumps({
    "pot_channels": power_of_two_channels,
    "pot_ms": round(pot_ms, 3),
    "npot_channels": non_power_of_two_channels,
    "npot_ms": round(npot_ms, 3),
    "onednn_flag": os.environ.get("TF_ENABLE_ONEDNN_OPTS", "default")
}))
```

```python
def run_channels_bench(env_extra):
    env = {**os.environ, "CUDA_VISIBLE_DEVICES": "-1", **env_extra}
    r = subprocess.run([sys.executable, "bench_channels.py"], env=env, capture_output=True, text=True, cwd=WORKSPACE)
    line = next((l for l in r.stdout.splitlines() if l.startswith("RESULT:")), None)
    if line is None:
        raise RuntimeError("Channel benchmark failed:\n" + r.stderr)
    return json.loads(line[len("RESULT:"):])

# Run the bench with oneDNN ON and OFF
results_off = run_channels_bench({"TF_ENABLE_ONEDNN_OPTS": "0"})
results_on = run_channels_bench({"TF_ENABLE_ONEDNN_OPTS": "1"})

# Calculate speedups
pot_speedup = results_off["pot_ms"] / results_on["pot_ms"]
npot_speedup = results_off["npot_ms"] / results_on["npot_ms"]

print("=" * 70)
print(" CONVOLUTIONAL CHANNEL ALIGNMENT BENCHMARK RESULTS")
print("=" * 70)
print(f"Power-of-Two Channels ({results_off['pot_channels']}):")
print(f"  • oneDNN OFF : {results_off['pot_ms']:.2f} ms")
print(f"  • oneDNN ON  : {results_on['pot_ms']:.2f} ms")
print(f"  • Speedup    : {pot_speedup:.2f}x")
print("-" * 70)
print(f"Non-Power-of-Two Channels ({results_off['npot_channels']}):")
print(f"  • oneDNN OFF : {results_off['npot_ms']:.2f} ms")
print(f"  • oneDNN ON  : {results_on['pot_ms']:.2f} ms")
print(f"  • Speedup    : {npot_speedup:.2f}x")
print("=" * 70)

import matplotlib.pyplot as plt

labels = ['Power-of-Two (128)', 'Non-Power-of-Two (123)']
speedups = [pot_speedup, npot_speedup]

plt.figure(figsize=(7, 4))
bars = plt.bar(labels, speedups, color=['#0071c5', '#00a3e0'], width=0.5)
plt.title("oneDNN Speedup by Channel Dimensional Alignment")
plt.ylabel("Speedup Factor (oneDNN ON vs OFF)")
plt.grid(axis='y', linestyle='--', alpha=0.7)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.05, f"{yval:.2f}x", ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```

### Activity 3.1: Automated Node-Type & Partition Suggester

We will write a functional script `suggest_partition.py` that implements a rule-based engine to evaluate hardware requirements. It analyzes target parameters such as model size (billions of parameters), batch size, and dataset scale to recommend either `xeon` (CPU) or `gaudi` (DL Accelerator) partitions, alongside identifying if the workload is likely **Compute-Bound** or **Memory-Bound**.

```python
%%writefile suggest_partition.py
import argparse
import sys

def evaluate_workload(model_params_b, batch_size, dataset_gb, precision):
    # Calculate estimated memory footprint in GB for model weights
    bytes_per_param = 4 if precision == "fp32" else (2 if precision in ("fp16", "bf16") else 1)
    model_memory_gb = model_params_b * bytes_per_param

    # Heuristic for partition selection
    # Gaudi is recommended for large models, massive datasets, or heavy training pipelines
    is_large_model = model_params_b >= 1.0
    is_large_dataset = dataset_gb >= 50.0

    if is_large_model or is_large_dataset:
        recommendation = "gaudi"
        reason = f"Model memory ({model_memory_gb:.2f} GB) or dataset size ({dataset_gb} GB) is large. Gaudi offers dedicated matrix acceleration."
    else:
        recommendation = "xeon"
        reason = f"Lightweight workload fits comfortably on standard scalable multithreaded Xeon nodes with lower orchestration overhead."

    # Estimate bottleneck
    # High batch sizes or massive models are typically Compute-bound on accelerators,
    # whereas small batch sizes with massive data streaming are often Memory/IO-bound.
    if batch_size >= 32 and model_params_b > 0.1:
        bottleneck = "Compute-Bound (Heavy matrix operations)"
    else:
        bottleneck = "Memory/IO-Bound (Data streaming or parameter fetching overhead)"

    return {
        "partition": recommendation,
        "reason": reason,
        "model_mem_est_gb": model_memory_gb,
        "bottleneck": bottleneck
    }

def main():
    parser = argparse.ArgumentParser(description="HPC Node-Type & Partition Suggester CLI")
    parser.add_argument("--model-size", type=float, default=0.5, help="Model size in Billions of Parameters")
    parser.add_argument("--batch-size", type=int, default=32, help="Target training/inference batch size")
    parser.add_argument("--dataset-size", type=float, default=10.0, help="Dataset size on disk in GB")
    parser.add_argument("--precision", type=str, choices=["fp32", "fp16", "bf16", "int8"], default="bf16", help="Compute precision")

    args = parser.parse_args()

    res = evaluate_workload(args.model_size, args.batch_size, args.dataset_size, args.precision)

    print("=" * 65)
    print(" HPC PARTITION & WORKLOAD RECOMMENDATION SYSTEM")
    print("=" * 65)
    print(f"[Input Workload Profile]:")
    print(f"  • Model Size      : {args.model_size} Billion parameters")
    print(f"  • Batch Size      : {args.batch_size}")
    print(f"  • Dataset Scale   : {args.dataset_size} GB")
    print(f"  • Precision       : {args.precision}")
    print("-" * 65)
    print(f"[Recommendation]  : --partition={res['partition'].upper()}")
    print(f"[Primary Reason]  : {res['reason']}")
    print(f"[Est. Model Mem]  : {res['model_mem_est_gb']:.2f} GB")
    print(f"[Est. Bottleneck] : {res['bottleneck']}")
    print("=" * 65)

if __name__ == "__main__":
    main()
```

```python
def partition_suggester(model_params_billions, dataset_size_gb, batch_size):
    """Advanced hardware mapping rules"""
    print(f"\nAnalyzing configuration: {model_params_billions}B Params, {dataset_size_gb} GB Data, Batch={batch_size}")
    if model_params_billions > 5.0:
        return "gaudi", "Model is too large for CPU; matrix math requires dedicated HBM2/3 Gaudi tensor engines."
    elif dataset_size_gb > 200.0 and batch_size < 32:
        return "xeon", "High-memory Xeon nodes with large shared system RAM are optimal for preprocessing heavy data loads."
    else:
        return "xeon", "General-purpose Xeon node is sufficient for smaller training configurations."

part, reason = partition_suggester(model_params_billions=7.0, dataset_size_gb=50, batch_size=8)
print(f"  • Recommended partition: --partition={part}\n  • Reason: {reason}")
```
